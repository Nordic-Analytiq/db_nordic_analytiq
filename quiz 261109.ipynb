```python
# A list of dictionaries representing server deployments
server_logs = [
    {"id": "SRV-101", "type": "Database", "port": 5432, "status": "active"},
    {"id": "SRV-102", "type": "Web", "port": 80, "status": "active"},
    {"id": "SRV-103", "type": "Cache", "port": 6379, "status": "inactive"},
    {"id": "SRV-104", "type": "Web", "port": 21, "status": "active"},       # Security Alert: Port 21 (FTP) is insecure!
    {"id": "SRV-105", "type": "Storage", "port": 443, "status": "active"},
    {"id": "SRV-106", "type": "Database", "port": 3306, "status": "active"}
]
healthy_servers = []
for log in server_logs:
    if log["status"] == "inactive":
        print(f"Status Alert: Server {log['id']} is inactive!")
        continue
    elif log["port"] == 21:
        print(f"CRITICAL: Server {log['id']} is running insecure protocol FTP on port 21!")
        continue
    else:
        healthy_servers.append(log['id'])
print(f"The total count of healthy servers discovered are: {len(healthy_servers)}")
# A list of healthy servers
print(f"The complete list of healthy server IDs are: {healthy_servers}")
```
