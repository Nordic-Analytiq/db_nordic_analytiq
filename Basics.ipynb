# **_Variables & Dynamic Typing_**

- The programs below shows how Python assigns data to variables without needing explicit type declarations, and how a single variable can change its data type on the fly (dynamic typing).

```python
name = "Anji"
age = 44
print("Hello, {name}. You are {age} years old.")
print(f"Hello, {name}. You are {age} years old.") #Modern Method (F-Strings)
print("Hello, " + name + ". You are " + str(age) + " years old.") #Old Method (Concatenation)

price = 100
tax_rate = 0.12

# Executes a math calculation inside the string
print(f"Total price including tax: ${price * (1 + tax_rate)}") # Output: Total price including tax: $112.00000000000001
print(f"Price: ${price}") # Output: Price: $100
# Formats numbers cleanly (e.g., forcing exactly 2 decimal places)
print(f"Total Formatted price including tax: ${price * (1 + tax_rate):.2f}") # Output: Total price including tax: $112.00
print(f"Formatted Price: ${price:.2f}") # Output: Formatted Price: $100.00
```

```python
# --- 1. Declaring Variables ---
# Python automatically figures out the data type based on the value assigned.

user_name = "Anji"          # String (str)
user_age = 44               # Integer (int)
account_balance = 1500.75   # Floating-point number (float)
is_active_member = True     # Boolean (bool)

print("--- Initial Variables ---")
print(f"Name: {user_name} (Type: {type(user_name)})")
print(f"Age: {user_age} (Type: {type(user_age)})")
print(f"Balance: ${account_balance} (Type: {type(account_balance)})")
print(f"Active: {is_active_member} (Type: {type(is_active_member)})")
print("-" * 30)

# --- 2. Demonstrating Dynamic Typing ---
# The same variable can point to a completely different data type later in execution.

data_holder = "It is a string string"
print(f"data_holder content: '{data_holder}' | Type: {type(data_holder)}")

data_holder = 42  # Now reassigned to an integer
print(f"data_holder content: {data_holder}   | Type: {type(data_holder)}")

data_holder = [1, 2, 3]  # Now reassigned to a list
print(f"data_holder content: {data_holder} | Type: {type(data_holder)}")
```

# **_Mandatory Whitespace Indentation_**

- The program below demonstrates how Python uses standard indentation (typically 4 spaces) instead of curly braces {} to define blocks of execution logic.

```python
# Try changing this value to test the different execution paths
temperature = 2

# The colon (:) tells Python that an indented code block is starting next.
if temperature > 30:
    # Everything indented by 4 spaces belongs inside this specific condition block.
    print("It is incredibly hot outside.")
    print("Make sure to turn on the air conditioning.")
elif temperature >= 15:
    # This block executes if the first condition is false, but this one is true.
    print("The weather is quite pleasant.")
    print("Perfect day for a walk outside.")
else:
    # This block executes if all previous conditions fail.
    print("It is quite chilly.")
    print("Don't forget your winter jacket.")

# This line has NO indentation, meaning it is outside the if/else structure.
# It will run every single time, regardless of what the temperature is.
print("Execution block finished.")
```

- Nested Indentation (Combining Logic)
- The program below shows how indentation stacks deeper when you nest blocks inside each other (e.g., an if statement placed inside a for loop)

```python
# A simple list of transaction amounts
transactions = [120.50, -45.00, 300.00, -12.99, 50.00]

print("Processing daily transactions...")

# Level 1 Indentation: Everything inside the loop is indented by 4 spaces
for amount in transactions:
    print(f"Evaluating amount: ${amount}")
    
    # Level 2 Indentation: Everything inside the conditional check is indented by 8 spaces
    if amount < 0:
        print("   -> Alert: This transaction is a withdrawal/expense.")
        print("   -> Updating outbound ledger.")
    else:
        print("   -> Notice: This transaction is a deposit/income.")
        print("   -> Updating inbound ledger.")

# Back to zero indentation: The loop has completely finished running
print("All transactions successfully processed.")
```

# **_Conditional Logic (if, elif, else)_**

The program below acts like a simple decision-making engine for an automated system. It uses elif (short for else if) to check multiple conditions sequentially from top to bottom.

```python
# Change this number to test different execution paths
account_balance = -50.00

print("Checking account status...")

if account_balance > 10000:
    print("💎 Status: Premium Tier Account.")
    print("   -> Enjoy priority customer support and 0% transfer fees.")

elif account_balance >= 100:
    print("✅ Status: Standard Tier Account.")
    print("   -> Account is active and in good standing.")

elif account_balance >= 0:
    print("⚠️ Status: Low Balance Warning.")
    print("   -> Please consider depositing funds soon to avoid service fees.")

else:
    # This executes only if the balance is strictly less than 0
    print("🚨 Status: Account Overdrawn!")
    print("   -> Transfer funds immediately to resolve the negative balance.")

print("Account check complete.")
```

# **_The for Loop (Iterating Over Collections)_**

- A for loop is used when you want to repeat an action a specific number of times or step through items in a collection (like a list) one by one.

```python
# Example A: Stepping through a list of text elements
cloud_services = ["Databricks", "Azure Data Factory", "Snowflake", "Synapse"]

print("--- Scanning Cloud Infrastructure Packages ---")
for service in cloud_services:
    # Python automatically assigns the current item to the variable 'service' on each loop
    print(f"Deploying and verifying component: {service}")

print("-" * 40)

# Example B: Using range() to repeat an action a set number of times
# range(1, 6) generates numbers from 1 up to (but not including) 6
print("--- Initiating Server Ping Countdown ---")
for seconds in range(1, 6):
    print(f"Ping attempt {seconds} in progress...")

print("System online!")
```

# **_The while Loop (Looping Until a Condition Changes)_**

- A while loop runs continuously as long as a specific condition remains True. It is ideal when you don't know exactly how many loops it will take to finish the job (e.g., waiting for a file download or a database connection).

```python
max_connection_retries = 3
current_attempt = 1
is_connected = False

print("Connecting to secure server cluster...")

# This loop keeps spinning as long as we haven't connected AND haven't run out of tries
while current_attempt <= max_connection_retries and not is_connected:
    print(f"Connecting... Attempt {current_attempt} of {max_connection_retries}")
    
    # Simulate an event: Let's assume connection succeeds only on attempt 3
    if current_attempt == 3:
        is_connected = True
        print("🤝 Secure handshake established successfully!")
    else:
        print("❌ Timeout detected. Retrying...")
    
    # CRITICAL: We must increment the counter manually. 
    # If we forget this line, the loop will run forever (an infinite loop)!
    current_attempt += 1

print("Network diagnostic execution complete.")
```

# **_Advanced Loop Control (break and continue)_**

Sometimes you need to alter the behavior of a loop while it is running:
- break: Exits the entire loop immediately, skipping everything else.
- continue: Skips the rest of the current loop cycle and jumps straight to the next item.

```python
# A list of data packets; 'CORRUPT' simulates a bad packet, 'END' is the end of the stream
data_stream = ["Packet_A", "Packet_B", "CORRUPT", "Packet_C", "END", "Packet_D"]

print("Analyzing data stream elements...")
for packet in data_stream:
    if packet == "CORRUPT":
        print("   ⚠️ Corrupt packet found! Skipping to the next one immediately.")
        continue  # Skips print(f"Processing...") and goes straight to Packet_C
        
    if packet == "END":
        print("   🏁 Termination signal encountered. Stopping stream analysis.")
        break     # Completely breaks out of the loop, skipping Packet_D entirely
        
    print(f"Successfully processed: {packet}")

print("Stream close protocol complete.")
```

## Built-in Data Structures

Python has four primary ways to group and organize data collections. Choosing the right one depends entirely on whether your data needs to be ordered, changeable, or unique.

# **_Lists([])_**

- What it is: A collection that is ordered, changeable, and allows duplicate values.
- When to use: Storing items where the specific sequence matters (like a timeline or queue).

```python
# Creating a list of data servers
servers = ["Azure_Node_1", "AWS_Node_2", "GCP_Node_3"]

# 1. Modifying elements and adding new ones
servers[0] = "Azure_Primary"    # Change an item using its index position (0-based)
servers.append("Hybrid_Node_4")  # Add an item to the very end

# 2. Accessing elements
print(f"First server in line: {servers[0]}")
print(f"Total active servers: {len(servers)}")

# 3. Iterating over a list
print("\nCurrent Server Cluster Layout:")
for server in servers:
    print(f" -> Active: {server}")
```

# **_Dictionaries ({key: value})_**

- What it is: A collection of key-value pairs that is ordered and changeable, but does not allow duplicate keys.
- When to use: Looking up information using a descriptive label rather than a number index (like a database record).

```python
# Storing configuration metrics using descriptive keys
database_config = {
    "host": "localhost",
    "port": 5432,
    "database": "production_db",
    "ssl_enabled": True
}

# 1. Accessing data via keys
print(f"Connecting to database: {database_config['database']} via port {database_config['port']}")

# 2. Modifying values and adding new key-value pairs
database_config["port"] = 5433           # Updates an existing key
database_config["timeout_seconds"] = 30  # Adds a completely brand new key

# 3. Iterating over keys and values safely
print("\nPrinting full system configurations:")
for key, value in database_config.items():
    print(f"  Configuration [ {key} ] = {value}")
```

# **_Tuples (())_**

- What it is: A collection that is ordered but completely unchangeable (immutable). It allows duplicate values.
- When to use: Storing constants or fixed records that should never accidentally be overwritten by your program while it's running (like GPS coordinates or fixed dimensions).

```python
# A tuple representing a fixed geospatial point (Latitude, Longitude)
datacenter_coordinates = (59.3293, 18.0686)

print(f"Datacenter Lat: {datacenter_coordinates[0]}")
print(f"Datacenter Long: {datacenter_coordinates[1]}")

# UNCOMMENTING THE LINE BELOW WILL CRASH THE PROGRAM:
# datacenter_coordinates[0] = 60.0000  
# TypeError: 'tuple' object does not support item assignment
```

# **_Sets ({})_**

- What it is: A collection that is unordered, unchangeable* (but you can add/remove items), and strictly unique (no duplicates allowed).
- When to use: Deduplicating data instantly or performing mathematical comparisons like intersections or unions.

```python
# A list with messy duplicate access logs
raw_logs = ["user_1", "user_2", "user_1", "user_3", "user_2", "user_4"]

# 1. Instantly stripping away duplicates by casting it into a set
unique_users = set(raw_logs)
print(f"Raw log count: {len(raw_logs)} | Unique user count: {len(unique_users)}")
print(f"Unique user set elements: {unique_users}")

# 2. Performing powerful set operations
admin_users = {"user_1", "user_5"}

# Find elements that exist in BOTH sets (Intersection)
active_admins = unique_users.intersection(admin_users)
print(f"Active users with admin privileges: {active_admins}")
```

- There are no hardcoded, structural limits built into Python that prevent you from casting collections back and forth (list -> set -> tuple).
- However, you will run into logical errors or resource barriers depending on the structure of your data and the size of your machine.
- To convert a list or tuple into a dictionary, the data must be structured as pairs. If Python cannot find exactly two elements per slot to map as a Key: Value pair, the conversion will crash instantly.

```python
# ✅ This works perfectly (Structured as pairs)
pairs_list = [("port", 5432), ("host", "localhost")]
config_dict = dict(pairs_list) 

# ❌ This crashes with a ValueError (No pair structure)
flat_list = ["port", 5432, "host", "localhost"]
config_dict = dict(flat_list)  
# Error: dictionary update sequence element #0 has length 4; 2 is required
```

- When you convert a list or tuple into a set, Python does not throw an error, but it permanently strips out duplicates and destroys the index order. If you convert it back to a list, your data might look completely different than it did originally.

```python
original = [3, 3, 2, 1, 2]
as_set = set(original)  # Becomes {1, 2, 3}
back_to_list = list(as_set)  # Becomes [1, 2, 3] -> Duplicates and order are gone!
```

If you are converting data sets containing millions of records, you will run into physical hardware barriers:
• Memory Spikes: During a conversion (like tuple(large_list)), Python temporarily keeps both objects in your RAM simultaneously while copying the elements over. If your original list takes up 4GB of RAM, converting it will cause your memory usage to spike to 8GB, which can crash your script if you run out of memory.
• Hashability: To put an item into a set or use it as a key in a dictionary, that item must be hashable (immutable). This means you cannot put a list inside a set or use a list as a dictionary key.

```python
# ❌ This will crash instantly
invalid_set = {, [3, 4] } 
# Error: TypeError: unhashable type: 'list'
```


Summary Checklist for Safe Conversions
- list ⇄ tuple: 100% safe at any time.
- list/tuple ➔ set: Safe, but you will lose duplicates and ordering.
- any ➔ dict: Only safe if the underlying data is nested perfectly in pairs of two.

```python
# We start with a changeable list of allowed IP addresses
allowed_ips_list = ["192.168.1.1", "10.0.0.1", "172.16.0.1"]

# The list can easily be altered right now
allowed_ips_list.append("192.168.1.99")
print(f"Original List: {allowed_ips_list} (Type: {type(allowed_ips_list)})")

# --- THE CONVERSION ---
# We use the tuple() function to freeze the data structure
final_security_policy = tuple(allowed_ips_list)

print(f"Converted Tuple: {final_security_policy} (Type: {type(final_security_policy)})")

# Trying to change the data now will fail and protect our infrastructure
# final_security_policy.append("8.8.8.8") # <-- This would crash the script!
```

```python
# A fixed, unchangeable tuple representing baseline server configurations
system_constants = ("CPU_Optimized", "RAM_64GB", "Storage_1TB")
print(f"Original Tuple: {system_constants} (Type: {type(system_constants)})")

# --- THE CONVERSION ---
# We use the list() function to unlock the data structure
editable_config = list(system_constants)

# Now we can safely modify it based on current application needs
editable_config.append("GPU_Enabled")
editable_config.remove("RAM_64GB")
editable_config.sort()

print(f"Converted & Modified List: {editable_config} (Type: {type(editable_config)})")
```


### Modifying an Item Inside a Tuple (The Workaround)
- Because tuples are strictly unchangeable, if you must update a specific item inside one, you have to use a "convert-modify-reconvert" sandwich loop:

```python
# A tuple of connection details where one port is wrong
db_connection_tuple = ("://server.com", "admin_user", 5432)
print(f"Old connection: {db_connection_tuple}")

# 1. Unlock it by turning it into a list
temp_list = list(db_connection_tuple)

# 2. Modify the incorrect element (Port 5432 -> 5433)
temp_list[2] = 5433

# 3. Freeze it back into a tuple
db_connection_tuple = tuple(temp_list)

print(f"Updated connection: {db_connection_tuple}")
```

Here is a comprehensive breakdown of all the standard operational methods available across Python’s four core data structures.

1. Lists ([]): The Versatile Workhorse.
Lists are mutable and ordered, meaning they have the largest selection of built-in manipulation methods.

Method	What it does	Example
- append(x)	Adds item x to the very end of the list.	[1, 2].append(3) → [1, 2, 3]
- insert(i, x)	Inserts item x at a specific index position i.	['a', 'c'].insert(1, 'b') → ['a', 'b', 'c']
- extend(iterable)	Appends all elements from another collection to the end.	[1].extend([2, 3]) → [1, 2, 3]
- remove(x)	Deletes the first occurrence of value x. Crashes if not found.	['a', 'b', 'a'].remove('a') → ['b', 'a']
- pop(i)	Removes and returns the item at index i. Defaults to last item.	x = [10, 20].pop() → x = 20, list becomes [10]
- clear()	Removes every single item, leaving the list empty.	[1, 2].clear() → []
- sort()	Sorts the list permanently in-place (ascending by default).	[3, 1, 2].sort() → [1, 2, 3]
- reverse()	Reverses the order of the items in-place.	[1, 2, 3].reverse() → [3, 2, 1]
- count(x)	Returns how many times value x appears in the list.	[1, 2, 1].count(1) → 2
- index(x)	Returns the index position of the first occurrence of x.	['a', 'b'].index('b') → 1


```python
# Comprehensive List Demo
tools = ["Spark", "Airflow"]
tools.append("Kafka")          # ['Spark', 'Airflow', 'Kafka']
tools.insert(1, "Databricks")   # ['Spark', 'Databricks', 'Airflow', 'Kafka']
tools.extend(["Trino", "Flink"])# ['Spark', 'Databricks', 'Airflow', 'Kafka', 'Trino', 'Flink']

tools.remove("Airflow")        # Removes Airflow
popped_item = tools.pop()      # Removes and returns 'Flink'
tools.sort(reverse=True)       # Sorts alphabetically backwards

print(f"Final List: {tools} | Popped: {popped_item}")
```

```python
tools = ["Spark", "Airflow", "Apache"]
popped_item = tools.pop(1)
print(f"Final List: {tools} | Popped: {popped_item}")
popped_item = tools.pop()
print(f"Final List: {tools} | Popped: {popped_item}")
```


2. Dictionaries ({}): Key-Value Pair Operations
Dictionaries are mutable, but because they rely on explicit key lookups instead of index tracks, they use a completely different set of methods. You cannot use append(), sort(), or insert() here.

Method	What it does	Example
- update(dict2)	Merges another dictionary or key-value pairs into this one.	d.update({"port": 80})
- pop(key)	Removes the specified key and returns its value.	val = d.pop("host")
- popitem()	Removes and returns the last inserted key-value pair as a tuple.	key, val = d.popitem()
- get(key, default)	Safely fetches a value. Returns the default if the key doesn't exist.	d.get("timeout", 30)
- keys()	Returns a dynamic view of all keys in the dictionary.	d.keys()
- values()	Returns a dynamic view of all values in the dictionary.	d.values()
- items()	Returns a dynamic view of all (key, value) pairs as tuples.	d.items()
- clear()	Completely empties the dictionary.	d.clear()


```python
# Comprehensive Dictionary Demo
metadata = {"engine": "Presto", "status": "active"}
print(f"Dict-metadata: {metadata}")
# Adding/Updating can be done directly or via update()
metadata["version"] = 3.1 
print(f"Dict-metadata: {metadata}")
metadata.update({"status": "running", "cluster": "eu-west"})
print(f"Dict-metadata: {metadata}")
# Safe fetching vs popping
cluster_region = metadata.get("cluster", "unknown")
print(f"Dict-metadata: {metadata}")
print(f"cluster_region: {cluster_region}")
removed_status = metadata.pop("status")
print(f"Dict-metadata: {metadata}")
print(f"Keys left: {list(metadata.keys())}")
print(f"Remaining Dict: {metadata}")
```

```python
# Comprehensive Dictionary Demo
metadata = {"engine": "Presto", "status": "active"}
print(f"Dict-metadata: {metadata}")
print(f"Dict-metadata items: {metadata.items()}")
print(f"Dict-metadata values: {metadata.values()}")
# Adding/Updating can be done directly or via update()
metadata["version"] = 3.1 
print(f"Dict-metadata: {metadata}")
metadata.update({"status": "running", "cluster": "eu-west"})
print(f"Dict-metadata: {metadata}")
cluster_region = metadata.get("cluster", "unknown")
print(f"Dict-metadata: {metadata}")
print(f"cluster_region: {cluster_region}")
metadata.popitem()
print(f"Dict-metadata: {metadata}")
cluster_region = metadata.get("cluster", "unknown")
print(f"Dict-metadata: {metadata}")
print(f"cluster_region: {cluster_region}")
```


3. Sets ({}): Mathematical & Unique Operations
Sets are mutable but completely unordered. Because there are no index numbers, methods like append(), insert(), sort(), or reverse() do not exist. Instead, they favor mathematical logic.

Method	What it does	Example
- add(x)	Adds a unique element x to the set. If it already exists, nothing changes.	{1}.add(2) → {1, 2}
- remove(x)	Deletes element x. Crashes with a KeyError if x is missing.	{1, 2}.remove(2) → {1}
- discard(x)	Deletes element x safely. If x is missing, it does nothing.	{1, 2}.discard(99) → {1, 2}
- pop()	Removes and returns an arbitrary (random) item from the set.	val = {5, 6}.pop()
- union(set2)	Returns a new set combining all unique items from both sets.	{1}.union({2}) → {1, 2}
- intersection(s2)	Returns a new set containing only items found in both sets.	{1, 2}.intersection({2, 3}) → {2}
- difference(s2)	Returns items that are in the first set but not in the second.	{1, 2}.difference({2, 3}) → {1}


```python
# Comprehensive Set Demo
monitored_nodes = {"node_A", "node_B"}
monitored_nodes.add("node_C")
monitored_nodes.add("node_A")  # Duplicate ignored silently

failed_nodes = {"node_B", "node_Z"}

# Intersections and Differences
both_monitored_and_failed = monitored_nodes.intersection(failed_nodes) # {"node_B"}
healthy_monitored_nodes = monitored_nodes.difference(failed_nodes)     # {"node_A", "node_C"}

print(f"Healthy Nodes: {healthy_monitored_nodes}")
```

While each data structure has specialized tricks, Python shares a core group of universal operators and built-in functions that work across all four structures (Lists, Dictionaries, Sets, and Tuples).
Here are the 6 universal operations that link them all together:

##1. The Membership Operator (in / not in)
Checks whether an element (or a key) exists inside the collection. This operation returns a boolean (True or False).

```python
# Works identically across all types
print("Spark" in ["Spark", "Kafka"])          # List -> True
print("host" in {"host": "localhost"})         # Dictionary (Checks Keys) -> True
print("node_A" in {"node_A", "node_B"})       # Set -> True
print(5432 in (5432, 80))                      # Tuple -> True
```

##2. Finding the Collection Size (len())

The global len() function evaluates any collection type and returns the total count of items contained inside it.

```python
print(len([10, 20, 30]))                        # List -> 3
print(len({"id": 1, "name": "Alex"}))          # Dictionary (Counts Pairs) -> 2
print(len({"apple", "orange"}))                # Set -> 2
print(len((5, 10, 15, 20)))                    # Tuple -> 4
```

Any core data structure can be dropped directly into a standard for loop to step through its elements one by one.

```python
# Lists, Sets, and Tuples loop over their individual values.
# Dictionaries loop over their KEYS by default.
for key in {"host": "localhost", "port": 5432}:
    print(f"Key found: {key}")
```

##4. Global Sorting Protection (sorted())

As mentioned briefly before, while you can't run .sort() on everything, you can pass any collection into the global sorted() function. Python will safely read the data, sort it, and output a brand new, sorted list without harming your original data structure.

```python
raw_set = {30, 10, 20}
new_sorted_list = sorted(raw_set)

print(new_sorted_list)  # Output: -> A standard list!
print(type(raw_set))      # Output: <class 'set'> -> Original set remains untouched

# Sample data: Student names and their exam scores
scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95, "Evan": 85}

# 1. Global sort by VALUES (Ascending: lowest to highest score)
sorted_by_value_asc = dict(sorted(scores.items(), key=lambda item: item[1]))
print("Globally sorted by value (Ascending):")
print(sorted_by_value_asc)
# Output: {'Charlie': 78, 'Alice': 85, 'Evan': 85, 'Bob': 92, 'Diana': 95}

# 2. Global sort by VALUES (Descending: highest to lowest score)
sorted_by_value_desc = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
print("\nGlobally sorted by value (Descending):")
print(sorted_by_value_desc)
# Output: {'Diana': 95, 'Bob': 92, 'Alice': 85, 'Evan': 85, 'Charlie': 78}

# 3. Global sort alphabetically by KEYS
sorted_by_key = dict(sorted(scores.items(), key=lambda item: item[0]))
print("\nGlobally sorted alphabetically by key:")
print(sorted_by_key)
# Output: {'Alice': 85, 'Bob': 92, 'Charlie': 78, 'Diana': 95, 'Evan': 85}

# Sample data
inventory = {"Laptop": 5, "Mouse": 50, "Keyboard": 25, "Monitor": 12}

# In-place sort by values (Ascending) using dictionary comprehension
inventory = {k: v for k, v in sorted(inventory.items(), key=lambda item: item[1])}

print("\nOriginal variable mutated/sorted in-place:")
print(inventory)
# Output: {'Laptop': 5, 'Monitor': 12, 'Keyboard': 25, 'Mouse': 50}

# Sort by value descending (highest score), then by key ascending (alphabetical tie-breaker)
# Note: To invert a string sort inside a lambda, we don't use a minus sign; 
# instead, we handle the criteria by feeding a tuple to the sorting key.
tie_broken_scores = dict(
    sorted(scores.items(), key=lambda item: (-item[1], item[0]))
)

print("\nSorted by score (descending), breaking ties alphabetically (ascending):")
print(tie_broken_scores)
# Output: {'Diana': 95, 'Bob': 92, 'Alice': 85, 'Evan': 85, 'Charlie': 78}
# (Alice comes before Evan because of 'A' vs 'E')

products = [
    {"name": "Log", "price": 10},
    {"name": "Anvil", "price": 150}
]

# MUST use lambda to extract the numeric price for comparison
cheapest = min(products, key=lambda x: x["price"])
print(f"The cheapest product is: {cheapest['name']} with a price of ${cheapest['price']}")

#Python has an optimized library called operator built for this. It runs faster than lambda because it executes entirely in C-code under the hood:
from operator import itemgetter
scores = {
    "Player_Alpha": 1250,
    "Player_Beta": 4200,
    "Player_Gamma": 950,
    "Player_Delta": 3100
}
# This replaces: key=lambda item: item[1]
sorted_dict = sorted(scores.items(), key=itemgetter(1))
print(sorted_dict)
```

##5. Finding Boundaries (min() and max())

If the collection contains comparable elements (like all numbers or all strings), you can pass the entire structure into min() or max().

```python
numeric_tuple = (45, 12, 89, 3)
print(min(numeric_tuple))  # Output: 3
print(max(numeric_tuple))  # Output: 89

stats = {"Alice": 85, "Bob": 92, "Charlie": 78}
# Find the key with the highest value
highest_scorer = max(stats, key=stats.get)
print(highest_scorer)

# Find the key with the lowest value
lowest_scorer = min(stats, key=stats.get)
print(lowest_scorer)

prices = {"apple": 0.50, "banana": 0.25, "orange": 0.75}

max_price = max(prices.values())
min_price = min(prices.values())

print(max_price)  # Output: 0.75
print(min_price)  # Output: 0.25

inventory = {"shoes": 12, "shirts": 45, "hats": 5}

# max() returns the (key, value) tuple with the highest value
most_stocked = max(inventory.items(), key=lambda item: item[1])
print(most_stocked)  
# Output: ('shirts', 45)

empty_dict = {}
result = max(empty_dict, default="No data") 
print(result) # Output: No data

data = {"a": 10, "b": None, "c": 25}
# Filter out None values before finding the max
clean_max = max((k for k, v in data.items() if v is not None), key=data.get)
print(clean_max) # Output: 'c'

employees = {
    "EMP01": {"name": "John", "salary": 50000},
    "EMP02": {"name": "Jane", "salary": 75000}
}

# MUST use lambda to drill down into the nested 'salary' key
highest_paid = max(employees.items(), key=lambda item: item[1]["salary"])
print(highest_paid)  # Output: ('EMP02', {'name': 'Jane', 'salary': 75000})

# Sample data: Usernames and their respective game scores
scores = {
    "Player_Alpha": 1250,
    "Player_Beta": 4200,
    "Player_Gamma": 950,
    "Player_Delta": 3100
}

# 1. Find the player with the HIGHEST score
top_player = max(scores, key=scores.get)

# 2. Find the player with the LOWEST score
lowest_player = min(scores, key=scores.get)

# Print results
print(f"The winner is: {top_player} (Score: {scores[top_player]})")
# Output: The winner is: Player_Beta (Score: 4200)

print(f"The lowest scorer is: {lowest_player} (Score: {scores[lowest_player]})")
# Output: The lowest scorer is: Player_Gamma (Score: 950)
```

##6. Clearing the Space (.clear() vs Variable Re-assignment)
Lists, Dictionaries, and Sets contain a native .clear() method to drop all values instantly while keeping the exact same structural object container active in memory.
Because Tuples are completely immutable, they do not have a .clear() method. To clear a tuple, you must completely re-assign the variable path to a fresh, empty tuple skeleton:

```python
# Mutable structures
my_list = [1, 2, 3]
my_list.clear()             # Becomes []

# Immutable structure workaround
my_tuple = (1, 2, 3)
my_tuple = ()               # Points the variable to a fresh empty tuple instead
```
