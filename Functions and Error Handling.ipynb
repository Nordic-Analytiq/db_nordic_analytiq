```python
# --- 1. Defining a Function ---
# 'def' tells Python you are building a function.
# 'server_id' and 'port' are parameters (inputs).
def check_security_threshold(server_id, port):
    """
    This is a docstring. It explains what the function does.
    Checks if a server port is within a dangerous range.
    """
    insecure_ports = [21, 23, 80]
    
    if port in insecure_ports:
        # 'return' sends a result back to whoever called the function
        return f"🚨 ALERT: {server_id} is using insecure port {port}!"
    else:
        return f"✅ {server_id} passed port security verification."

# --- 2. Calling the Function ---
# We pass arguments into the function and capture the returned text.
result_1 = check_security_threshold("SRV-101", 5432)
result_2 = check_security_threshold("SRV-104", 21)

print(result_1)
print(result_2)
```

```python
# A list containing a string that cannot be converted to a number
raw_ports = ["80", "443", "UNKNOWN_PORT", "8080"]

print("Starting port parser pipeline...")

for raw_val in raw_ports:
    try:
        # Python will try to execute this block line-by-line
        parsed_port = int(raw_val)
        print(f"Successfully converted port token to integer: {parsed_port}")
        
    except ValueError:
        # If a 'ValueError' happens inside the try block, execution instantly jumps here
        print(f"❌ DATA ERROR: Could not convert '{raw_val}' to a valid integer. Skipping row.")
        
    except Exception as e:
        # A catch-all for any other unexpected errors (like system failures)
        print(f"💥 UNKNOWN CRASH: An unexpected error occurred: {e}")

print("Pipeline execution completed without crashing!")
```

```python
def simulated_database_query(query_string):
    print("\n[System] Opening connection pool to PostgreSQL...")
    try:
        if "DROP TABLE" in query_string:
            # Manually triggering a security error for demonstration
            raise PermissionError("Write access denied on production layer.")
        print(f"[System] Query executed successfully: '{query_string}'")
        
    except PermissionError as e:
        print(f"[Security Alert] Intercepted Violation: {e}")
        
    finally:
        # This will execute regardless of whether the query succeeded or threw a PermissionError
        print("[System] CRITICAL CLEANUP: Safely closing database connection pool.")

# Test Path A: Runs completely safely
simulated_database_query("SELECT * FROM users;")

# Test Path B: Catches error but still triggers the cleanup block smoothly
simulated_database_query("DROP TABLE users;")
```

```python
import os

# --- Function 1: Writing Data to a File ---
def save_security_report(filename, report_lines):
    try:
        # 'w' mode will create the file if it doesn't exist, or overwrite it
        with open(filename, 'w', encoding='utf-8') as file:
            for line in report_lines:
                # We manually add a newline character (\n) to push text to the next line
                file.write(line + "\n")
        print(f"✅ Success: Report cleanly written to '{filename}'.")
    except IOError as e:
        print(f"❌ Disk Error: Could not write file structure to disk. Details: {e}")


# --- Function 2: Reading Data From a File ---
def display_security_report(filename):
    print(f"\n--- Reading Content From: {filename} ---")
    try:
        # 'r' mode opens the file strictly for reading
        with open(filename, 'r', encoding='utf-8') as file:
            # Method A: Reading line-by-line using a loop (highly memory efficient)
            for line in file:
                # strip() removes trailing newlines so print doesn't double-space
                print(line.strip())
                
    except FileNotFoundError:
        print(f"❌ Error: The file '{filename}' does not exist on this machine.")
    except IOError as e:
        print(f"❌ Read Error: System failed to extract data bytes. Details: {e}")


# ==========================================
# --- Executing the File I/O Operations ---
# ==========================================

target_file = "cloud_audit_log.txt"
mock_log_data = [
    "TIMESTAMP: 2026-11-09 | NODE: SRV-101 | EVENT: Port Sweep Passed",
    "TIMESTAMP: 2026-11-09 | NODE: SRV-104 | EVENT: Port 21 Blocked",
    "TIMESTAMP: 2026-11-09 | NODE: SRV-106 | EVENT: DB Backup Executed"
]

# 1. Try to read a file that doesn't exist yet to test error handling
display_security_report("missing_file.txt")

# 2. Write our logs safely to the disk
save_security_report(target_file, mock_log_data)

# 3. Read the successfully generated log file back into memory
display_security_report(target_file)

# Clean up the file from your computer after the test (Optional script safety)
if os.path.exists(target_file):
    #os.remove(target_file)
    print(f"\n[Cleanup] Temp file '{target_file}' cleanly removed from disk environment.")
```
