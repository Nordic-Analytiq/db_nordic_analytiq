# Lecture - MLflow for GenAI

## Overview
Before you build and register a retrieval pipeline, you need the tooling that tracks, traces, and governs it. This lecture introduces MLflow's generative AI capabilities — tracking, tracing, model flavors, and the Model Registry — and shows how Unity Catalog provides unified governance for models alongside your data.

## Learning Objectives
By the end of this lecture, you will be able to:
1. Describe MLflow's Gen AI features.
2. Register a RAG pipeline to the Model Registry.
