# Demo - Prompt Engineering in AI Playground

## Overview
This lesson introduces the foundational concepts of prompt engineering and then puts them into practice in the Databricks AI Playground. You will first learn what a prompt is, the components of an effective prompt, and techniques for guiding model behavior. You will then use the AI Playground to see how prompt design influences the accuracy and reliability of generative AI responses, with a focus on identifying and mitigating hallucinations.

## Learning Objectives
By the end of this demo, you will be able to:
- Define the basics of prompt engineering and describe the components of an effective prompt.
- Explain basic prompt engineering techniques for guiding model responses.
- Describe the features of the Databricks AI Playground.
- Apply techniques to construct prompts that guide generative AI models to provide responses without hallucinations.
- Augment prompts with additional context to improve the accuracy and relevance of generated responses.

## B. Prompt Engineering Tips & Tricks

### B1. Tip: Prompts are Model-specific

#### A prompt guides the model to complete task(s)

- Different models may require **different prompts**.
- Provide **examples and cues** to guide model’s response generation.
- Different **use cases** may require different prompts techniques.
    -   Example: Classification task vs. translation
- **Iterative development** is key.
    -   Iterate by adjusting the **temperature parameter**—higher for more creative outputs, lower for focused responses.
- Be aware of **biased** responses.
- Check for **hallucinations** and **inaccurate information**.

### B3. Tip: Guide the model for better responses

- **Ask the model not to make things up/hallucinate**
    - Do not make things up if you do not know
        - Say, 'I do not have that information'

- **Ask the model not to assume or probe for sensitive information**
    - Do not make assumptions based on nationalities
    - Do not ask the user to provide their SSNs

- **Ask the model not to rush to a solution**
    - Ask it to take more time to “think” (Chain-of-Thought for reasoning)
    - Explain how you solve this math problem
    - Do this step-by-step
        - Step 1: Summarize into 100 words
        - Step 2: Translate from English to French

## G. Hallucination and Context: A Fictional Example

This section demonstrates how prompt design and context affect the likelihood of hallucination in large language model (LLM) responses. We use a fictional scenario to illustrate three approaches:

1. A prompt that encourages hallucination
2. A prompt that discourages hallucination
3. A prompt augmented with additional context

Each example shows how prompt engineering can guide model behavior and output quality.

### G1. Simple Prompt Which Hallucinates

When prompted for information about a fictional person or event, LLMs may confidently generate plausible but entirely invented details—especially if the system prompt encourages creativity or gap-filling. This example demonstrates how a model can hallucinate when asked about a person who does not exist. Paste the system prompt and query below into the AI Playground to observe the behavior.

### G3. Augment Prompt with Additional Context

Providing relevant context in the prompt—similar to Retrieval Augmented Generation (RAG)—guides the model to generate more accurate and grounded responses. In this example, we supply detailed fictional achievements for Alex Johnson, which the model uses to produce a biography aligned with the provided information.

**💬 Query:**
> Provide a detailed biography of Alex Johnson, the first human ambassador to the Galactic Federation, highlighting their key diplomatic achievements and the impact on Earth-Galactic relations in the 23rd century.
> 
> **Context:**
> Alex Johnson (born 2201, New Auckland, Earth) is a career diplomat and linguist fluent in seven Earth languages and three Galactic Federation dialects. Appointed as Earth's first official ambassador to the Galactic Federation in 2247 by the United Earth Council, Johnson served two consecutive terms until 2261.
>
> Key achievements:
> - **Earth-Galactic Technology Pact (2249):** Johnson led eighteen months of negotiations that resulted in the Technology Pact, granting Earth access to Galactic Federation propulsion and medical technologies in exchange for Earth's rich biological diversity data.
> - **Andromeda–Alpha Centauri Trade Dispute (2253):** Acting as neutral mediator, Johnson brokered the Centauri Accord, resolving a decade-long trade embargo between the Andromedan Collective and the Alpha Centauri Commerce Union, opening new trade routes that benefited Earth's interstellar economy.
> - **Earth Seat on the Galactic Senate (2258):** Johnson's advocacy secured Earth a permanent observer seat on the Galactic Senate, the first step toward full membership.
>
> Johnson retired in 2262 and was awarded the United Earth Medal of Honour and the Galactic Federation's Star of Diplomacy — the first human recipient of the latter.

**🤖 Response:**
> (Model generates a biography closely following the provided context.)

This demonstrates how context injection can mitigate hallucination and enhance output quality, emphasizing the importance of context in RAG systems.

## Conclusion

In this lesson, you learned the fundamentals of prompt engineering—what a prompt is, its components, and techniques for guiding model behavior—and then applied them in the Databricks AI Playground to understand and mitigate hallucinations in generative AI models.

Key takeaways:
- A well-formed prompt combines an instruction, context, an input/question, and an output format.
- Prompt design is model-specific and iterative; formatting and clear guidance improve results.
- Poorly designed prompts can lead to confident but incorrect (hallucinated) responses.
- Clear instructions in system prompts can reduce hallucination and improve reliability.
- Augmenting prompts with relevant context—manually or via RAG—can further ground model outputs.
