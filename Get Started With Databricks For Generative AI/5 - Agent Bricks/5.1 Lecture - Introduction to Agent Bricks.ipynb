# Lecture - Introduction to Agent Bricks

## Overview
In the previous modules you built, evaluated, and deployed a RAG pipeline by hand. This lecture introduces **Agent Bricks**, the Databricks approach to building agents that are auto-optimized on your data. You will see the challenges Agent Bricks addresses and the key task types it offers: Information Extraction, Knowledge Assistant, Supervisor Agent, and Custom LLM.

## Learning Objectives
By the end of this lecture, you will be able to:
1. Discuss the challenges of building GenAI systems.
2. Explain how Agent Bricks addresses these challenges.
3. Define core features of Agent Bricks.
4. Distinguish the key Agent Bricks task types and their common use cases.
