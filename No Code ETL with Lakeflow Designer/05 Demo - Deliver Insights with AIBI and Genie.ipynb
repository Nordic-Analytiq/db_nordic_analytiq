### Genie Agent (Space): two ways to deliver it

**Default dashboard Genie Agent (Space) - the option we use here**

- Comes with every published AI/BI dashboard. When you publish, the **Enable Genie** toggle is on by default.
- Databricks auto-generates it from the dashboard's own published datasets and visualizations, so there is nothing to curate.
- Republishing the dashboard automatically refreshes the agent to match the new datasets and visuals. No manual upkeep.
- Best when the ask is simply "let viewers chat with the data behind this dashboard." Viewers click **Ask Genie** right on the dashboard.

**Standalone Genie Agent (Space) - not used in this demo**

- A purpose-built, reusable chat experience that lives on its own, independent of any single dashboard.
- You explicitly curate it: choose the datasets, add example SQL queries, write instructions and business context, and control sharing.
- Best when you want one trusted natural-language interface across data assets that many people reuse, or when you need to tune answer quality.

**Good to know - limits of the dashboard agent**

- Since we are using the default dashboard agent in this demo, it is worth knowing what it will not do.
- The biggest one: you can open it, but you cannot edit its instructions in the Genie UI. If you need fully curated, editable context, that is when you reach for a standalone agent instead.
- It also stays hidden from the Genie listing page and file browser, and caps at 30 dashboard datasets.
- For the full list of limitations, see the AI/BI dashboard Genie Agent documentation.

**Quick rule of thumb**

- Just want questions answered about this dashboard? Use the built-in default (what we do here).
- Need a governed, reusable Q&A experience you shape yourself? Build a standalone agent.

<details>

### See everything in Genie One

This is the final step of the demo. Up to now you have been the builder. Here you switch seats and experience your work the way a business user would, all from one place: Genie One.

**Genie One is the single entry point for business users**

- It is the simplified UI where consumers interact with data and AI without touching compute, queries, or notebooks.
- It is not a single feature or a linear set of steps. It is a home. From it, users fan out to whatever they need: the assets shared with them, a natural-language question, or personalized recommendations.

**Find and open your work**

- Genie One surfaces the assets shared with you: AI/BI dashboards, Genie Agents (Spaces), Databricks Apps, and saved queries, tables, and documents.
- The dashboard you published and the Genie Agent tied to it both appear here, alongside everything else the consumer has access to.
- Click an asset to open it in its familiar experience (for example, a dashboard opens as a dashboard).

**Ask in natural language**

- Genie One includes a unified chat. Ask a plain-English question, for example "Which topic had the most attendees?" and it answers from your data.
- Chat searches across your dashboards, queries, and Genie Agents (Spaces) to find the answer, so the consumer never writes SQL.
- You can also connect external sources like Google Drive or SharePoint to bring your company documents into the conversation.

**Discover what is relevant**

- The "For you" section recommends assets based on your activity and what is trending with similar users.
- Domains let you browse assets grouped by business context instead of navigating catalog hierarchies.
- Favorites and recent items make it easy to get back to what you use most.

**Why this matters**

- It closes the loop from raw data to a governed gold layer, to a dashboard and agent, and finally to a self-service home where a business user can find answers.
- The takeaway for learners: once the gold tables, dashboard, and agent exist, delivering insights is a matter of finding or asking, not building.

</details>
