# No Code ETL with Lakeflow Designer

No-Code ETL with Lakeflow Designer shows how to build a complete medallion ETL workflow, from bronze to silver to gold, visually and without writing pipeline code. You will use Databricks Lakeflow Designer and its built-in AI authoring assistant, Genie Code.

In Lakeflow Designer, you build workflows on a visual canvas using operators. Operators are the building blocks of a data pipeline, with each operator performing an action such as adding a source, filtering or transforming data, aggregating data, joining datasets, or creating an output table.

Throughout the course, you will build workflows in two ways:

- Add and configure operators using the point-and-click interface.
- Describe transformations in natural language and use Genie Code to generate operators and workflows.

Using a realistic meeting and session dataset that includes sessions, participants, and feedback, the course introduces the medallion architecture in Unity Catalog and shows how to add bronze sources to the canvas, clean and reshape data into silver tables, and aggregate and join data into analytics-ready gold tables.

You will also learn how to schedule the workflow and deliver insights to business users through an AI/BI dashboard, a Genie Agent (Space), and Genie One. An optional bonus topic extends the workflow with the AI Function operator to derive sentiment from free-form feedback.

By the end of the course, you will understand how raw data moves through a governed and scheduled medallion architecture to produce analytics-ready gold tables and consumer-ready insights using a visual, AI-assisted experience on the Databricks Lakehouse.

## Terminal Objectives

By the end of this course, you will be able to:
- Describe the medallion architecture (bronze to silver to gold) and how Unity Catalog and Catalog Explorer are used to inspect source data, schemas, and join keys.
- Explain how the source and silver layers are built in Lakeflow Designer using multiple authoring styles (point-and-click operators, natural-language Genie Code, and the SQL operator), including flattening semi-structured JSON.
- Describe how analytics-ready gold tables are created by aggregating and joining silver tables, how the results are validated, and how a workflow is scheduled from the Designer UI (including when to use Lakeflow Jobs for broader orchestration).
- Explain how insights are delivered to business users with an AI/BI dashboard, a Genie Agent (Space), and Genie One, and when to use each.
- Deliver insights via AI/BI dashboard, Genie Agent (Space), and Genie One
- Describe how to use the AI Function operator (`ai_sentiment`) to gain insights from free-form text feedback.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **Basic familiarity with the Databricks workspace.** You can navigate the workspace UI, open and use notebooks, and attach a SQL warehouse or other compute resource.

- **Basic familiarity with Unity Catalog.** You understand the three-level namespace (catalog, schema, table) and can find a table in the catalog explorer.

- **Working knowledge of SQL, including joins.** You can read and write SELECT statements with WHERE, GROUP BY, and aggregates, and you can join two tables on a shared key.
