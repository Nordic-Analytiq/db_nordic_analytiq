```python
def findNumber(arr, k):
    """Return 'YES' if k exists in arr, otherwise 'NO'."""
    return 'YES' if k in arr else 'NO'

# Test with a sample array
arr = [1, 3, 5, 7, 9, 11]
print('arr =', arr)
print('findNumber(arr, 5)  ->', findNumber(arr, 5))   # exists -> YES
print('findNumber(arr, 4)  ->', findNumber(arr, 4))   # missing -> NO
print('findNumber(arr, 11) ->', findNumber(arr, 11))  # exists -> YES
print('findNumber(arr, 2)  ->', findNumber(arr, 2))   # missing -> NO
```

```python
def odd_numbers_between(start, end):
    """Return the list of odd numbers between start and end (inclusive)."""
    lo, hi = min(start, end), max(start, end)
    return [n for n in range(lo, hi + 1) if n % 2 != 0]

# Test with sample inputs
print(odd_numbers_between(3, 15))   # -> [3, 5, 7, 9, 11, 13, 15]
print(odd_numbers_between(8, 20))    # -> [9, 11, 13, 15, 17, 19]
print(odd_numbers_between(20, 8))    # reversed order still works
print(odd_numbers_between(4, 4))     # no odd numbers in range

# Interactive version: enter two numbers when prompted
# a = int(input('Enter first number: '))
# b = int(input('Enter second number: '))
# print('Odd numbers:', odd_numbers_between(a, b))
```

```python
# Python lists are 0-indexed: the first element is at position 0
arr = [10, 20, 30, 40, 50]

print('arr        =', arr)
print('arr[0]     =', arr[0])    # first element
print('arr[-1]    =', arr[-1])   # negative index counts from the END
print('arr[1:4]   =', arr[1:4])  # slice: start at 1, stop BEFORE 4 (same rule as range!)
print('arr[:3]    =', arr[:4])   # omitted start = beginning
print('arr[::2]   =', arr[::2])  # step of 2 = every other element
```

```python
# LISTS are mutable: you can change individual elements in place
nums = [1, 2, 3]
nums[0] = 99
print('list after nums[0] = 99 :', nums)   # works -> [99, 2, 3]

# STRINGS are immutable: they can never be changed after creation
word = 'hello'
# word[0] = 'H'   # <- would raise TypeError: 'str' object does not support item assignment

# To "change" a string you build a NEW one
new_word = 'H' + word[1:]
print('rebuilt string             :', new_word)
```

```python
# You often need BOTH the position and the value while looping
# Anti-pattern: for i in range(len(arr))
# Pythonic way: enumerate() yields (index, value) pairs

arr = [5, 12, 9, 20, 3]
arr_enumerated= enumerate(arr)
print(arr)
print(list(arr))
#print(arr_enumerated) --this statement throw error becasue enumerated is a list so list should 
print(list(arr_enumerated))
for i, value in enumerate(arr):
    kind = 'odd' if value % 2 != 0 else 'even'   # same test you used in odd_numbers_between
    print(f'position {i}: value {value} is {kind}')

# enumerate can also start counting at any number
print(list(enumerate(['a', 'b', 'c'], start=2)))
```

```python
# A list comprehension builds a list in one expression:
#   [new_value for item in iterable if condition]
# You already used one: [n for n in range(lo, hi+1) if n % 2 != 0]

squares      = [n * n for n in range(1, 6)]        # transform each item
evens_only   = [n for n in range(1, 11) if n % 2 == 0]  # filter items
odd_labels   = ['odd' if n % 2 else 'even' for n in range(1, 6)]  # conditional value

print('squares    :', squares)
print('evens_only :', evens_only)
print('odd_labels :', odd_labels)
```

```python
# A dict stores key -> value pairs; lookup by key is instant (hash table)
ages = {'puli': 30, 'ada': 36, 'linus': 54}

print('ages            :', ages)
print("ages['ada']     :", ages['ada'])        # direct lookup
print("ages.get('zoe') :", ages.get('zoe'))       # missing key -> None (no crash)
print("ages.get('zoe', 0):", ages.get('zoe', 0))  # missing key -> your default

ages['zoe'] = 25        # add a new pair
print('after adding zoe:', ages)

# Looping gives you keys by default; .items() gives (key, value) pairs
for name, age in ages.items():
    print(f'{name} is {age}')
```

```python
# 'in' works on both lists and sets, but the COST is very different:
#   list  -> checks elements one by one   -> O(n)
#   set   -> hash lookup, jumps straight to it -> O(1)
import time

big_list = list(range(200_000))   # 0..199,999
big_set  = set(big_list)
target = 199_999                  # worst case: last element of the list

start = time.perf_counter()
_ = target in big_list
t_list = time.perf_counter() - start

start = time.perf_counter()
_ = target in big_set
t_set = time.perf_counter() - start

print(f'list membership: {t_list * 1000:.3f} ms')
print(f'set  membership: {t_set * 1000:.3f} ms')
print(f'list was ~{t_list / t_set:.0f}x slower here')
```

```python
# Functions can have DEFAULT values and be called with KEYWORD arguments
def greet(name, greeting='Hello', punct='!'):
    return f'{greeting}, {name}{punct}'

print(greet('Puli'))                          # defaults fill the gaps
print(greet('Puli', 'Welcome'))               # positional: fills in order
print(greet('Puli', punct='?'))               # keyword: names the parameter
print(greet(punct='.', greeting='Hey', name='Ada'))  # keywords can be any order

# Watch out: mutable defaults are shared across calls (classic Python gotcha)
def append_bad(item, bucket=[]):
    bucket.append(item)
    return bucket
print(append_bad(1), append_bad(2))   # [1, 2] <- the SAME list persists!
```

```python
# Errors (exceptions) can be CAUGHT so the program keeps running
# e.g. user input that is not a number:

for raw in ['42', 'abc', '7.5', '13']:
    try:
        n = int(raw)
        print(f'{raw!r:>6} -> int {n}, {"odd" if n % 2 else "even"}')
    except ValueError:
        print(f'{raw!r:>6} -> not a whole number, skipped')

# try/except/else/finally: else runs when NO error occurred,
# finally runs ALWAYS (cleanup, e.g. closing a file)
try:
    value = int('100')
except ValueError:
    print('conversion failed')
else:
    print('conversion ok ->', value)
finally:
    print('finally: this always runs')
```

```python
# f-strings embed expressions directly inside text: f'...{expr}...'
name, score, total = 'Puli', 17, 20

print(f'{name} scored {score}/{total}')
print(f'percentage : {score / total:.1%}')      # format as percentage, 1 decimal
print(f'padded     : [{score:5d}]')               # width 5, right-aligned
print(f'expression : {score * 3 + 1}')            # any expression works
print(f'debug form : {score=}')                   # prints name AND value
print(f"quotes: it's easy to nest {'quotes' if True else ''}")
```

```python
# sorted() returns a NEW sorted list; .sort() sorts in place
nums = [42, 7, 19, 3]
print('sorted(nums)          :', sorted(nums))
print('sorted(nums, reverse=True):', sorted(nums, reverse=True))
print('original unchanged     :', nums)

# key= tells Python WHAT to sort by - run on each element first
people = [('ada', 36), ('puli', 30), ('zoe', 25), ('linus', 54)]
print('by name (item 0)       :', sorted(people, key=lambda p: p[0]))
print('by age  (item 1)       :', sorted(people, key=lambda p: p[1]))
print('oldest to youngest     :', sorted(people, key=lambda p: -p[1]))
print('names longer than 3, sorted by length:',
      sorted([n for n, _ in people if len(n) > 3], key=len))
```
